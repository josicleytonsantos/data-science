#### 📘 O que é o Scikit-learn?

O **Scikit-learn** (`sklearn`) é uma biblioteca de **aprendizado de máquina em Python** 🐍, construída sobre ferramentas como NumPy e SciPy.

👉 Ele permite aplicar modelos de Machine Learning **sem precisar implementar toda a matemática por trás**.

Com ele, é possível trabalhar com:

* 📈 Regressão
* 🧩 Classificação
* 📊 Clusterização
* 🔻 Redução de dimensionalidade

#### 🤖 O que é Aprendizado de Máquina?

Aprendizado de Máquina é uma área da Inteligência Artificial que permite **criar modelos que aprendem padrões a partir de dados** 📊

Antes:

* Usávamos Pandas para tratar dados
* Calculávamos estatísticas (média, etc.)

Agora:

* Queremos **prever valores futuros** 🔮

#### 📉 Tipos de Modelos Lineares

##### 📈 Regressão Linear

* Prever valores numéricos
* Ex: número de clientes

[
y = ax + b
]

👉 `a` = inclinação
👉 `b` = intercepto

##### 🔀 Classificação Linear

* Separar dados em classes
* Ex: “compra” vs “não compra”

#### 📚 Bibliotecas Utilizadas

````python
import numpy as np              # 🔢 Operações numéricas (arrays)
import matplotlib.pyplot as plt # 📊 Gráficos
import seaborn as sns           # 🎨 Visualizações mais bonitas e alguns datasets
import pandas as pd             # 🧾 Manipulação de dados
````

#### ⚙️ 1) Instanciando o Modelo

````python
from sklearn import linear_model  # Importa modelos lineares

# Cria o modelo de regressão linear
regressao_linear = linear_model.LinearRegression()
````

#### 🧠 2) Treinando o Modelo (fit)

##### 📌 Conceito

* **X (entrada)** → variáveis (features)
* **Y (saída)** → resultado esperado

````python
# Temperatura (entrada)
X = [[20], [21], [25], [29], [31], [32], [32], [34]]

# Número de clientes (saída)
Y = [4, 2, 7, 15, 16, 14, 15, 20]

# Visualizando os dados
plt.scatter(X, Y)
````

````python
# Treinando o modelo
regressao_linear.fit(X, Y)

# Coeficiente angular (inclinação da reta)
a = regressao_linear.coef_

# Intercepto (onde corta o eixo Y)
b = regressao_linear.intercept_

print(a, b)
# Exemplo: y = 1.16x - 20.85
````

#### 🔮 3) Fazendo Previsões

````python
# Novas temperaturas
X2 = [[40], [41], [18]]

# Prevendo clientes
previsoes = regressao_linear.predict(X2)

print(previsoes.astype(int))  # Converte para inteiro (opcional)

# `.predict()` usa o modelo treinado para estimar novos valores.
````

#### 📊 Visualizando o Modelo

````python
fig, ax = plt.subplots()

# Dados reais
ax.scatter(X, Y)

# Linha da regressão
ax.plot(X, regressao_linear.predict(X))

plt.show()
````

#### 🌐 Regressão Multidimensional

Agora usamos **mais de uma variável** 🧠

Exemplo:

* Entrada → `[displacement, weight]`
* Saída → `horsepower

##### 📊 Visualização 3D

````python
fig = plt.figure(layout='tight')
ax = fig.add_subplot(projection='3d')

mpg = sns.load_dataset('mpg').dropna()

ax.set_xlabel('displacement')
ax.set_ylabel('weight')
ax.set_zlabel('horsepower')

ax.scatter(mpg['displacement'], mpg['weight'], mpg['horsepower'])

# Cada ponto representa um carro 🚗
````

##### 🧾 Preparando os dados

````python
# Entradas (duas variáveis)
X = np.array(mpg[['displacement', 'weight']])

# Saída
Y = np.array(mpg['horsepower'])

# Visualizando parte dos dados
display(mpg[['displacement', 'weight']].head())
display(mpg['horsepower'].head())
````

##### ⚙️ Treinando

````python
regressao_linear = linear_model.LinearRegression()

# Treinamento
regressao_linear.fit(X, Y)

# Previsão
Y_previsto = regressao_linear.predict(X)

# Agora o modelo é: y = a_1x_1 + a_2x_2 + b
````

##### 📊 Visualizando o plano

````python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# Pontos previstos
ax.scatter(X[:,0], X[:,1], Y_previsto, c='red')
````

#### 🔀 Classificação com Scikit-learn

Na **classificação**, queremos prever **categorias** (rótulos), e não números.

👉 Estrutura dos dados:

* **X (entrada)** → lista de vetores (features)
* **Y (saída)** → lista de categorias

##### 🐧 Exemplo: Classificando pinguins

Cada entrada pode ser:

```text
x = [peso, grossura do bico]
y = 'macho' ou 'fêmea'
```

📌 Ou seja:

* Entradas → características físicas
* Saída → classe (sexo do pinguim)

#### ⚙️ Modelo utilizado

Vamos usar um classificador linear do Scikit-learn:

```python
from sklearn import svm

# Criando o classificador
classificador = svm.LinearSVC()
```

💡 O `LinearSVC` tenta encontrar uma **reta (ou plano)** que separa as classes.

#### 🧠 Treinamento e Previsão

```python
# Treinando o modelo
classificador.fit(X, Y)

# Fazendo previsões
classificador.predict([[4000, 18]])
```

📌 Aqui estamos perguntando:

> “Com peso 4000g e bico 18mm, é macho ou fêmea?”

#### 📊 Avaliando o Modelo

```python
taxa = classificador.score(X, Y)

print(f'Taxa de acerto: {round(100*taxa, 2)}%')
```

👉 O método `.score()`:

* Compara previsões com valores reais
* Retorna a **acurácia** (taxa de acertos)

⚠️ Importante:

* Isso está sendo feito nos **mesmos dados de treino**
* Pode dar uma taxa **otimista demais** (overfitting)

#### 📦 Dataset (Seaborn)

Usando dataset pronto do Seaborn:

````python
penguins = sns.load_dataset('penguins').dropna()

# Entradas: [grossura do bico, peso]
X = np.array(penguins[['bill_depth_mm', 'body_mass_g']])

# Saída: sexo
Y = np.array(penguins['sex'])
````

#### 🎨 Visualizando os dados

````python
# Cores por classe
cores = ['red' if s == 'Female' else 'blue' for s in penguins['sex']]

# Scatter plot
plt.scatter(penguins['body_mass_g'], penguins['bill_depth_mm'], c=cores)

# Visualizando dados
display(penguins[['bill_depth_mm', 'body_mass_g']].head())
display(penguins['sex'].head())

# Vermelho = fêmea
# Azul = macho
````

#### ⚙️ Treinando e avaliando

````python
from sklearn import svm

classificador = svm.LinearSVC()

# Treinamento
classificador.fit(X, Y)

# Avaliação
taxa = classificador.score(X, Y)

print(f'Taxa de acerto: {round(100*taxa, 2)}%')
````

#### 📈 Visualizando a fronteira de decisão

````python
fig, ax = plt.subplots()

# Plot dos dados
ax.scatter(penguins['body_mass_g'], penguins['bill_depth_mm'], c=cores)

# Coeficientes do modelo
w = classificador.coef_      # pesos
b = classificador.intercept_ # bias

# Equação da reta
a = -(w[0][1] / w[0][0])
b = -b[0] / w[0][0]

# Linha de separação
x_vals = np.linspace(2500, 6500, 100)
y_vals = a * x_vals + b

ax.plot(x_vals, y_vals, c='black')

'''

#### 🧠 O que está acontecendo aqui?

* O modelo aprende uma equação do tipo:

[
w_1x_1 + w_2x_2 + b = 0
]

* Isso define uma **reta separadora** 📏
* Essa reta divide:

  * um lado → machos
  * outro lado → fêmeas
  
'''
````
